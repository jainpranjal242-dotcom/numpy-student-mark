```python
import numpy as np
marks = np.array([78, 85, 92, 67, 88, 74, 95, 81, 69, 90])
marks
```

```python
mean = np.mean(marks)
median = np.median(marks)
maximum = np.max(marks)
minimum = np.min(marks)
std = np.std(marks)
print('Mean:', mean)
print('Median:', median)
print('Maximum:', maximum)
print('Minimum:', minimum)
print('Standard Deviation:', std)
```

## Result Explanation
- Mean: average of all marks.
- Median: middle value after sorting.
- Maximum: highest mark.
- Minimum: lowest mark.
- Standard deviation: spread of marks around the mean.
